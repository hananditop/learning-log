# Module 1: Descriptive Statistics

descriptive statistics = way to **summarize data** with numbers and graphs (figures)

why descriptive statistics?
space shuttle Challenger (1986) crashed because of the 29F temperature (not likely, because usually it launchs at 60s, 70s, 80s F)

function:
- communicate information (tell you what)
- support reasoning (the why)

ways to summarize data:
1. numerical (some number things) summary
2. graphical summary
but it is better to use graphical summary because people prefer look at pictures than numbers

for *qualitative* data (ex: color, car types, origin, etc) = **pie chart, dot plot**. both correspond percentage (%)

advantages:
- compare 2 categories = dot plot, because it easier to judge the position in horizontal line
- see what % of total category represent = pie chart

for *quantitative* data (ex: numbers) = **bar graph**. because:
- the ordering
- the distance between number

**histogram** = use block with different width (range), areas are proportional to frequency, total area corresponds to 100%

kinds of information in histogram:
1. density (crowding): how many subjects (people) for 1 unit, example: the highest density is around age 19 as 4%
2. percentages: area = height x width, example: 14% of all subjects/people fall into 60-80: 20 years (range between 60-80) x 0.7% (the density)

**boxplot** (box-and-whisker plot) = visualize 5 key numbers:
1. lower whisker = the smallest number (10 miles per gallon)
2. upper whisker = the largest number (35 miles per gallon)
3. median = half smaller and half larger (19 miles per gallon)
4. first quartile = quarter smaller and 3-quarter larger
5. third quartile = 3-quarter smaller and quarter larger

boxplot is less information (only 5), but it's less space

**scatterplot** = not one but two data (pairs), example: education and salaries for number of people, useful to visualize the relationship

provide context using *small multiples* to help compare patterns across groups

3D bars are quite difficult to read, but grid lines help, better to use dotplot

summarizing data with one number = **mean** or **median**

mean = average

median = half are larger and half are smaller (mid)

mean = median when the histogram is symmetric

*skewed to the right* = right side is longer than left side (mean > median)

example: if median sales price of 10 homes is 1M, we know that 5 homes sold for 1M or more

<=1M <=1M <=1M <=1M 1M (median) >=1M >=1M >=1M >=1M >=1M

if average (mean) sales price of 10 homes is $1M, we can't draw a conclusion

example: sum of 10 sales prices = 1M,
if one sold for 8M, then average sales of other 9 homes = 2M/9 = 200.000

if the histogram is skewed, better to use median, knowing average doesn't tell you much

**percentiles** = 90th percentile means 90% of households report an income of 135.000 or less, only 10% more (you are the top 10% of rich people if you are in the 90th percentile with more than 135.000)
- 1st quartile = 25th percentile
- 2nd quartile = 50th percentile (median)
- 3rd quartile = 75th percentile

in boxplot, distance between 3rd quartile and 1st quartile tells how much the data spread out = **interquartile range**

measuring spread = **standard deviation**

x-bar = average number of x1, x2, ..., xn

s = squareroot(1/n * sigma (xi - x-bar)**2))

interpretation: difference of each number (xi) from its average (x-bar), squared it, divided by total numbers, square root

square root of average of the difference of each numbers from its average

sometime n - 1 instead of n

x-bar and s are used to summarize data, both are sensitive to few or small data
- x-bar = measure of center
- s = measure of spread

if very skewed, use median and interquartile range
