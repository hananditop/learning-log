# Module 5: Conditional Code Notes

## Part 1

```python
# <, <=, ==, =>, >, !=

# comparison operators
x = 5
if x == 5:
    print("Equals 5")
if x > 4:
    print("Greater than 4")
if x >= 5:
    print("Greater than or Equals 5")
if x < 6: print("Less than 6")
if x <= 5:
    print("Less than or Equals 5")
if x != 6:
    print("Not equal 6")
```

```python
# one-way decisions
x = 5
print("Before 5")
if x == 5:
    print("Is 5")
    print("Is Still 5")
    print("Third 5")
print("Afterwards 5")
print("Before 6")
if x == 6:
    print("Is 6")
    print("Is Still 6")
    print("Third 6")
print("Afterwards 6")
```

```python
# nested decisions
x = 42
if x > 1:
    print("More than one")
    if x < 100:
        print("Less than 100")
print("All done")
```

```python
# two-way decisions
x = 4
if x > 2:
    print("Bigger")
else:
    print("Not bigger")
print("All done")
```

## Part 2

```python
# multi-way
if x < 2:
    print("small")
elif x < 10:
    print("Medium")
else:
    print("LARGE")
print("All done")
```

```python
# multi-way puzzles
# Which will never print regardless of the value for x?

if x < 2:
    print("Below 2")
elif x >= 2:
    print("Two or more")
else:
    print("Something else")
# Something else

if x < 2:
    print("Below 2")
elif x < 20:
    print("Below 20")
elif x < 10:
    print("Below 10")
else:
    print("Something else")
# Below 10
```

## Part 3

```python
# try / except

astr = "Hello Bob"
try:
    istr = int(astr)
except:
    istr = -1
    
print("First", istr)

astr = 123
try:
    istr = int(astr)
except:
    istr = -1
    
print("Second", istr)
```

```python
astr = "Bob"
try:
    print("Hello")
    istr = int(astr)
    print("There")
except:
    istr = -1

print("Done", istr)
```

```python
# sample try / except
rawstr = input("Enter a number: ")
try:
    ival = int(rawstr)
except:
    ival = -1

if ival > 0:
    print("Nice work")
else:
    print("Not a number")
```
