Program steps/flow:
- sequential: do one thing, then one thing, then one thing
- conditional: if the thing is right, do it, if wrong, skip it and else do the other thing
- repeated
- store a step to be used over and over

```python
# sequential
x = 1
print(x)
x = x + 1
print(x)
```

```python
# conditional
x = 5
if x < 10:
    print("Smaller")
if x > 20:
    print("Bigger")
print("Finis")
```

```python
# repeated
n = 5
while n > 0:
    print(n)
    n = n-1
print("Blastof!")
```
